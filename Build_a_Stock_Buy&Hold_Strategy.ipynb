```python
import pandas as pd
```

```python
data = pd.read_csv("msft_5y.csv", index_col = ["Date"])

data.head()
```

```python
data.tail()
```

```python
data.index = pd.to_datetime(data.index)
```

```python
data.head()
```

```python
data = data.asfreq("ME")

data.head()
```

```python
data = data.dropna()

data.head()
```

```python
percent_change = data.pct_change()

percent_change.head()
```

```python
percent_change = percent_change[["Open", "High", "Low", "Close", "Volume"]]

percent_change.head()
```

```python
percent_change = percent_change.dropna()

percent_change.head()
```

```python
returns = percent_change.mean(axis = 1)

returns.head()
```

```python
total_returns = (returns + 1).cumprod()

total_returns.head()
```

```python
import matplotlib.pyplot as pyplot

pyplot.rcParams["figure.figsize"] = (10, 10)
pyplot.rcParams.update({"font.size" : 22})

pyplot.ylabel("Returns")

total_returns.plot()
```
