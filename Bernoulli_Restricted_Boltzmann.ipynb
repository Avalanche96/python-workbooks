```python
import yfinance

Stock_Symbol = "BTC-USD"

Start_Date = "2020-01-01"

End_Date = "2025-01-01"

dataframe = yfinance.download(tickers = Stock_Symbol,
                             start = Start_Date,
                             end = End_Date)
dataframe
```

```python
import numpy as np

Decreased = 1

Increased = 0

dataframe["Volume Increase/Decrease"] = np.where(dataframe["Volume"].shift(-1) > dataframe["Volume"],
                                                Decreased,
                                                Increased)

dataframe
```

```python
dataframe["Returns"] = dataframe["Close"].pct_change()

dataframe
```

```python
dataframe = dataframe.dropna()

dataframe
```

```python
Buy = 1

Sell = 0

dataframe["Buy/Sell"] = np.where(dataframe["Close"].shift(-1) > dataframe["Close"],
                                Buy,
                                Sell)
dataframe
```

```python
dataframe["High vs Low"] = (dataframe["High"] - dataframe["Low"])/dataframe["Low"]

dataframe
```

```python
dataframe["Open vs Close"] = (dataframe["Open"] - dataframe["Close"])/dataframe["Open"]

dataframe
```

```python
X = np.array(dataframe.drop(["Buy/Sell"], axis = 1))

y = np.array(dataframe["Buy/Sell"])

X
```

```python
y
```

```python
from sklearn.neural_network import BernoulliRBM

Number_Of_Components = 2

neural_network = BernoulliRBM(n_components = Number_Of_Components)

neural_network.fit(X)
```

```python
neural_network.score_samples(X)
```

```python
neural_network.transform(X)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_regression = LogisticRegression()

neural_network_2 = BernoulliRBM(n_components = Number_Of_Components)

from sklearn.pipeline import Pipeline

classifier = Pipeline(steps = [("neural_network", neural_network_2), ("logistic_regresssion", logistic_regression)])
```

```python
from sklearn.model_selection import train_test_split

Seed = 42

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = Seed)
```

```python
X_train
```

```python
classifier.fit(X_train, y_train)
```

```python
predictions = classifier.predict(X_test)

predictions
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test, predictions)
```
