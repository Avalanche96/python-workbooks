```python
import yfinance

stock_symbol = "AMD"
start_date = "2014-01-01"
end_date = "2024-01-01"

dataframe = yfinance.download(tickers = stock_symbol,
                             start = start_date,
                             end = end_date)

dataframe
```

```python
import numpy

Buy = 1
Sell = 0

dataframe["Buy or Sell on Close"] = numpy.where(dataframe["Close"].shift(-1) > dataframe["Close"], Buy, Sell)

dataframe
```

```python
dataframe["Return"] = dataframe["Close"].pct_change()

dataframe
```

```python
Volume_Decreased = 0
Volume_Increased = 1

dataframe["Volume Increase/Decrease"] = numpy.where(dataframe["Volume"].shift(-1) > dataframe["Volume"], Volume_Decreased, Volume_Increased)

dataframe
```

```python
dataframe = dataframe.dropna()

dataframe
```

```python
dataframe["Buy or Sell on Open"] = numpy.where(dataframe["Open"].shift(-1) > dataframe["Open"], Buy, Sell)

dataframe
```

```python
X = dataframe.drop(["Buy or Sell on Close"], axis = 1)

y = dataframe["Buy or Sell on Close"]

X
```

```python
y
```

```python
X = X.reset_index(drop = True)
y = y.reset_index(drop = True)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)

X_train
```

```python
y_train.shape
```

```python
X_test.shape
```

```python
y_test.shape
```

```python
from sklearn.ensemble import RandomForestClassifier

Max_Number_of_Trees = 50

model = RandomForestClassifier(n_estimators = Max_Number_of_Trees)

model.fit(X_train, y_train)

prediction = model.predict(X_test)

from sklearn import metrics

print(metrics.accuracy_score(y_test, prediction))
```

```python
X.columns = ['_'.join(col) if isinstance(col, tuple) else col for col in X.columns]
```

```python
Importances = model.feature_importances_
```

```python
import pandas as pd

feature_importance_df = pd.DataFrame({'Feature' : X.columns.tolist(), 'Importance' : Importances})

feature_importance_df = feature_importance_df.sort_values(by='Importance', ascending=False)
```

```python
import matplotlib.pyplot as pyplot

figure, axes = pyplot.subplots(figsize = (20, 10))

import seaborn
seaborn.barplot(x = 'Importance', y = 'Feature', data = feature_importance_df)

pyplot.title('Feature Importance')

pyplot.show()
```
