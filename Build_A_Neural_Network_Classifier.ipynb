```python
import yfinance

Stock_Symbol = "AMD"
Start_Date = "2020-01-01"
End_Date = "2025-01-01"

dataframe = yfinance.download(tickers = Stock_Symbol,
                             start = Start_Date,
                             end = End_Date)

dataframe
```

```python
dataframe["Returns"] = dataframe["Close"].pct_change()

dataframe
```

```python
dataframe = dataframe.dropna()
```

```python
dataframe
```

```python
import numpy as np

Buy = 1
Sell = 0

dataframe["Buy/Sell on Open"] = np.where(dataframe["Open"].shift(-1) > dataframe["Open"],
                                       Buy,
                                       Sell)
dataframe
```

```python
dataframe["Buy/Sell on Close"] = np.where(dataframe["Close"].shift(-1) > dataframe["Close"],
                                       Buy,
                                       Sell)
dataframe
```

```python
Decreased = 1
Increased = 0

dataframe["Volume Increase/Decrease"] = np.where(dataframe["Volume"].shift(-1) > dataframe["Volume"],
                                                Decreased,
                                                Increased)
dataframe
```

```python
y = dataframe["Buy/Sell on Close"]

X = dataframe.drop(["Buy/Sell on Close"], axis = 1)

y
```

```python
X
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y)

X_train
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

scaler
```

```python
scaler.fit(X_train)
```

```python
X_train = scaler.transform(X_train)

X_train
```

```python
X_test = scaler.transform(X_test)

X_test
```

```python
y_train
```

```python
from sklearn.neural_network import MLPClassifier

Side_Length = 12
Maximum_Epochs = 300

neural_network = MLPClassifier(hidden_layer_sizes = (Side_Length, Side_Length, Side_Length), max_iter = Maximum_Epochs)

neural_network.fit(X_train, y_train)
```

```python
predictions = neural_network.predict(X_test)

predictions
```

```python
from sklearn.metrics import confusion_matrix

confusion_matrix(y_test, predictions)
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, predictions))
```
