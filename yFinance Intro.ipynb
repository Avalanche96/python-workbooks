```python
import yfinance as yf
import pandas as pd

symbols = ['AMD', 'INTC', 'AOSL']
symbolData = []

for symbol in symbols:
    stock = yf.Ticker(symbol)
    symbolInfo = stock.info
    pd.set_option('display.max_rows', None)
    ##how to find various data from Yahoo Finance
    insiderPurchases = stock.insider_purchases
    ##print(insiderPurchases)
    balanceSheet = stock.balance_sheet.iloc[:,0]
    ##print(balanceSheet)
    totalEquity = balanceSheet.get('Stockholders Equity')
    totalLiabilities = balanceSheet.get('Total Liabilities Net Minority Interest')
    totalAssets = balanceSheet.get('Total Assets')
    totalCurrentAssets = balanceSheet.get('Current Assets')
    totalCurrentLiabilities = balanceSheet.get('Current Liabilities')
    Inventory = balanceSheet.get('Inventory')
    ##print(totalCurrentAssets)
    ##debt to equity
    if totalLiabilities and totalEquity:
        de_ratio = totalLiabilities/totalEquity
    else:
        de_ratio = None
    ##current ratio
    if totalCurrentAssets and totalCurrentLiabilities:
        CurrentRatio = totalCurrentAssets / totalCurrentLiabilities
    else:
        CurrentRatio = None    
    ##quick ratio
    if totalCurrentAssets and totalCurrentLiabilities:
        QuickRatio = (totalCurrentAssets - Inventory) / totalCurrentLiabilities
    else:
        QuickRatio = None
    symbol_data = {
        "Ticker" : symbol,
        "Total Equity" : totalEquity,
        "Total Liabilities" : totalLiabilities,
        "Total Assets" : totalAssets,
        "Total Current Assets" : totalCurrentAssets,
        "Total Current Liabilities" : totalCurrentLiabilities,
        "Inventory" : Inventory,
        "Debt to Equity Ratio" : de_ratio,
        "Current Ratio" : CurrentRatio,
        "Quick Ratio" : QuickRatio,
        "TrailingPE" : symbolInfo.get('trailingPE'),
        "PEG Ratio" : symbolInfo.get('pegRatio'),
        "EV/EBITDA" : symbolInfo.get('enterpriseToEbitda'),
        "ROE" : symbolInfo.get('returnOnEquity')
    
    }
    symbolData.append(symbol_data)


for stock in symbolData:
    print("\nStock Information")
    for key, value in stock.items():
        print(f"{key}: {value}")
                  
```
