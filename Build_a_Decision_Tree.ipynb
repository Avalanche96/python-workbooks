```python
import yfinance

Stock_ticker = "AMD"
Start_date = "2019-01-01"
End_date = "2024-01-01"

dataframe = yfinance.download(tickers = Stock_ticker,
                 start = Start_date,
                end = End_date)
dataframe
```

```python
import numpy

Buy = 1
Sell = 0

dataframe["Buy or Sell"] = numpy.where(dataframe["Close"].shift(-1) > dataframe["Close"], Buy, Sell)

dataframe
```

```python
dataframe["Buy or Sell on Open"] = numpy.where(dataframe["Open"].shift(-1) > dataframe["Open"], Buy, Sell)

dataframe
```

```python
dataframe["Returns"] = dataframe["Close"].pct_change()

dataframe
```

```python
dataframe = dataframe.dropna()

dataframe
```

```python
Volume_Drop = 1
Volume_Increase = 0

dataframe["Volume Increase or Decrease"] = numpy.where(dataframe["Volume"].shift(-1) > dataframe["Volume"], Volume_Drop, Volume_Increase)

dataframe
```

```python
X = dataframe.drop(["Buy or Sell"], axis = 1)

X
```

```python
y = dataframe["Buy or Sell"]

y
```

```python
X.info()
```

```python
y.info()
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(random_state = 0)

model = model.fit(X, y)

from sklearn import tree

visualization_data = tree.export_graphviz(model, out_file = None, feature_names = X.columns, class_names = ["Buy", "Sell"])

visualization_data
```

```python
import pydotplus

decision_tree_graph = pydotplus.graph_from_dot_data(visualization_data)

decision_tree_graph
```
