```python
numerical_features = ["SalePrice", "LotArea", "OverallQual", "OverallCond", "YearBuilt", "1stFlrSF", "2ndFlrSF", "BedroomAbvGr"]

categorical_features = ["MSZoning", "LotShape", "Neighborhood", "CentralAir", "SaleCondition", "MoSold", "YrSold"]

dataframe = dataframe[numerical_features + categorical_features]
print(dataframe)
```

```python
dataframe["AgeWhenSold"] = dataframe["YrSold"] - dataframe["YearBuilt"]

numerical_features.append("AgeWhenSold")

dataframe["AgeWhenSold"].hist(edgecolor= "Black", bins=20)
```

```python
dataframe["MSZoning"].value_counts().plot(kind="bar", title="MSZoning")
```

```python
import matplotlib.pyplot as pyplot

figure, axis = pyplot.subplots(4, 2, figsize=(20, 20))

for feature, subplot in zip(categorical_features, axis.flatten()):
    dataframe[feature].value_counts().plot(kind="bar", ax=subplot, title=feature)
```

```python
def identify_above30(series):

    counts = series.value_counts()

    return list(counts[counts >= 30].index)

categories_to_keep = dataframe[categorical_features].apply(identify_above30, axis=0)

for feature in categorical_features:

    dataframe = dataframe.loc[dataframe[feature].isin(categories_to_keep[feature])]

dataframe.shape
```

```python
figure, axis = pyplot.subplots(4, 2, figsize=(20, 20))

for feature, subplot in zip(categorical_features, axis.flatten()):
    dataframe[feature].value_counts().plot(kind="bar", ax=subplot, title=feature)
```

```python
sorted_months_sold = dataframe.groupby("MoSold")["SalePrice"].median().sort_values().index.values

figure, axis = pyplot.subplots(figsize = (14, 14))

seaborn.boxplot(x = "MoSold", y = "SalePrice", data = dataframe, order = sorted_months_sold, ax = axis)
```
