```python
import pandas as pd

tables = pd.read_html("https://en.wikipedia.org/wiki/List_of_S%26P_500_companies", header = 0)

tables
```

```python
table = tables[0]

table
```

```python
symbol_column = table["Symbol"]

symbol_column
```

```python
stock_tickers_list = list(symbol_column)

stock_tickers_list
```

```python
import yfinance

start_date = "2023-01-01"
end_date = "2024-01-01"

dataframe = yfinance.download(tickers = stock_tickers_list, start = start_date, end = end_date)["Close"]

dataframe
```

```python
returns = dataframe.pct_change().mean() * 252

returns = pd.DataFrame(returns)

returns
```

```python
returns.columns = ["Returns"]

returns
```

```python
Trading_Days_Per_Year = 252

import math

returns["Volatility"] = dataframe.pct_change(fill_method = None).std() * math.sqrt(Trading_Days_Per_Year)

returns
```

```python
import numpy as np

data = np.asarray([np.asarray(returns["Returns"]), np.asarray(returns["Volatility"])]).T

data
```

```python
Null_Replacement = 0

cleaned_data = np.where(np.isnan(data), Null_Replacement, data)

cleaned_data
```

```python
X = cleaned_data
```

```python
from sklearn.cluster import KMeans

Min_Clusters = 1
Max_Clusters = 30
Max_Iterations = 500
distortions = []

for number_of_clusters in range(Min_Clusters, Max_Clusters + 1):
    model = KMeans(n_clusters = number_of_clusters, init = "k-means++", n_init = 10, max_iter = Max_Iterations, random_state = 0)
    model.fit(X)
    distortions.append(model.inertia_)

distortions
    
```

```python
import matplotlib.pyplot as pyplot

pyplot.figure(figsize = (20, 8))

pyplot.plot(range(Min_Clusters, Max_Clusters + 1), distortions, marker = 'o')

pyplot.xlabel("Number of Clusters")

pyplot.ylabel("Distortion (Inertia)")

pyplot.show()
```

```python
from scipy.cluster.vq import vq, kmeans

Number_Of_Clusters = 10

centroids, distortion_kmeans = kmeans(cleaned_data, Number_Of_Clusters)

idx, distortion_vq = vq(cleaned_data, centroids)

idx
```
